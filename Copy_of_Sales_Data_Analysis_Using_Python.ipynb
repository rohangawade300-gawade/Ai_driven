```python

import pandas as pd
import matplotlib.pyplot as plt

# Create sample sales data
data = {
    "Product": ["Laptop", "Mobile", "Headphones",
                "Laptop", "Mobile", "Headphones",
                "Laptop", "Mobile", "Headphones"],
    "Month": ["Jan", "Jan", "Jan",
              "Feb", "Feb", "Feb",
              "Mar", "Mar", "Mar"],
    "Units_Sold": [10, 25, 40, 15, 30, 45, 12, 35, 50],
    "Price": [50000, 15000, 2000,
              50000, 15000, 2000, 50000, 15000, 2000]
}

df = pd.DataFrame(data)

# Calculate revenue
df["Revenue"] = df["Units_Sold"] * df["Price"]

print("SALES DATA")
print(df.to_string(index=False))

# Total revenue by product
product_sales = df.groupby("Product")["Revenue"].sum()

print("\nTotal Revenue by Product:")
print(product_sales)

# Total revenue by month
monthly_sales = df.groupby("Month")["Revenue"].sum()
month_order = ["Jan", "Feb", "Mar"]
monthly_sales = monthly_sales.reindex(month_order)

print("\nMonthly Revenue:")
print(monthly_sales)

# Best-selling product by units
units = df.groupby("Product")["Units_Sold"].sum()
print("\nBest-selling Product:", units.idxmax())
print("Units Sold:", units.max())

# Product revenue chart
product_sales.plot(kind="bar", title="Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Monthly revenue chart
monthly_sales.plot(kind="bar", title="Monthly Sales Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.tight_layout()
plt.show()
```

```python
print("PROJECT INSIGHTS")

print("1. Highest Revenue Product:",
      product_sales.idxmax())

print("2. Highest Revenue:",
      "₹", product_sales.max())

print("3. Best Selling Product:",
      units.idxmax())

print("4. Best Sales Month:",
      monthly_sales.idxmax())

print("5. Total Revenue:",
      "₹", df["Revenue"].sum())
```
